```python
import pandas as pd
from sklearn.impute import SimpleImputer

df_clean = df_copy.copy(deep=True)

df_clean = df_clean.sort_values(
    ["Patient_ID", "ICULOS"]
).reset_index(drop=True)

numerical_cols = [
    col for col in df_clean.select_dtypes(include="number").columns
    if col != "SepsisLabel"
]

df_clean[numerical_cols] = (
    df_clean.groupby("Patient_ID")[numerical_cols].ffill()
)

imputer = SimpleImputer(strategy="median")
df_clean[numerical_cols] = imputer.fit_transform(
    df_clean[numerical_cols]
)

print("Total missing values:", df_clean.isnull().sum().sum())
print("Dataset shape:", df_clean.shape)

display(df_clean.head())
```

```python
print(df_clean["SepsisLabel"].value_counts())
print("\nTarget percentages:")
print(df_clean["SepsisLabel"].value_counts(normalize=True) * 100)
```

```python
from sklearn.model_selection import GroupShuffleSplit

drop_cols = ["SepsisLabel", "Patient_ID"]

X = df_clean.drop(
    columns=[col for col in drop_cols if col in df_clean.columns]
)
y = df_clean["SepsisLabel"]
groups = df_clean["Patient_ID"]

split1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.30,
    random_state=42
)

train_idx, temp_idx = next(split1.split(X, y, groups=groups))

X_temp = X.iloc[temp_idx]
y_temp = y.iloc[temp_idx]
groups_temp = groups.iloc[temp_idx]

split2 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=42
)

val_rel_idx, test_rel_idx = next(
    split2.split(X_temp, y_temp, groups=groups_temp)
)

val_idx = temp_idx[val_rel_idx]
test_idx = temp_idx[test_rel_idx]

X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
X_test, y_test = X.iloc[test_idx], y.iloc[test_idx]

print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)

train_patients = set(groups.iloc[train_idx])
val_patients = set(groups.iloc[val_idx])
test_patients = set(groups.iloc[test_idx])

print("Train-validation overlap:", len(train_patients & val_patients))
print("Train-test overlap:", len(train_patients & test_patients))
print("Validation-test overlap:", len(val_patients & test_patients))
```

```python
print("Training labels:")
print(y_train.value_counts())

print("\nValidation labels:")
print(y_val.value_counts())

print("\nTesting labels:")
print(y_test.value_counts())
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import (
    classification_report,
    average_precision_score,
    roc_auc_score
)

neg = (y_train == 0).sum()
pos = (y_train == 1).sum()
scale_weight = neg / pos

print("Scale positive weight:", scale_weight)

xgb_model = XGBClassifier(
    n_estimators=600,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_weight,
    eval_metric="aucpr",
    random_state=42,
    n_jobs=-1
)

# Train model
xgb_model.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    verbose=50
)

y_val_prob = xgb_model.predict_proba(X_val)[:, 1]
y_val_pred = (y_val_prob >= 0.5).astype(int)

print("\nValidation Classification Report:")
print(classification_report(y_val, y_val_pred, zero_division=0))

print("Validation AUPRC:", average_precision_score(y_val, y_val_prob))
print("Validation ROC-AUC:", roc_auc_score(y_val, y_val_prob))
```

```python
import pandas as pd
from sklearn.model_selection import ParameterGrid
from sklearn.metrics import average_precision_score
from xgboost import XGBClassifier

param_grid = {
    "max_depth": [3, 4, 5],
    "learning_rate": [0.03, 0.05],
    "min_child_weight": [1, 5]
}

results = []
best_model = None
best_auprc = -1
best_params = None

for params in ParameterGrid(param_grid):

    model = XGBClassifier(
        n_estimators=1000,
        max_depth=params["max_depth"],
        learning_rate=params["learning_rate"],
        min_child_weight=params["min_child_weight"],
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_weight,
        eval_metric="aucpr",
        early_stopping_rounds=50,
        random_state=42,
        n_jobs=-1,
        tree_method="hist"
    )

    model.fit(
        X_train,
        y_train,
        eval_set=[(X_val, y_val)],
        verbose=False
    )

    val_prob = model.predict_proba(X_val)[:, 1]
    val_auprc = average_precision_score(y_val, val_prob)

    results.append({
        **params,
        "Best_iteration": model.best_iteration,
        "Validation_AUPRC": val_auprc
    })

    print(
        f"Parameters: {params} | "
        f"Best iteration: {model.best_iteration} | "
        f"Validation AUPRC: {val_auprc:.5f}"
    )

    if val_auprc > best_auprc:
        best_auprc = val_auprc
        best_model = model
        best_params = params

results_df = pd.DataFrame(results).sort_values(
    "Validation_AUPRC", ascending=False
)

print("\nBest Hyperparameters:", best_params)
print("Best Validation AUPRC:", round(best_auprc, 5))
print("Best Boosting Iteration:", best_model.best_iteration)

display(results_df)

model_path = r"C:\Users\kanam\OneDrive\Documents\project\PhysioNet2019\xgboost_tuned_model.pkl"

joblib.dump(best_model, model_path)

print("Best model saved successfully!")
print("Saved at:", model_path)
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

train_ids = df_clean.loc[X_train.index, "Patient_ID"]
val_ids = df_clean.loc[X_val.index, "Patient_ID"]
test_ids = df_clean.loc[X_test.index, "Patient_ID"]

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train).astype(np.float32)
X_val_scaled = scaler.transform(X_val).astype(np.float32)
X_test_scaled = scaler.transform(X_test).astype(np.float32)

print("Train:", X_train_scaled.shape)
print("Validation:", X_val_scaled.shape)
print("Test:", X_test_scaled.shape)
```

```python
def create_sequences(X, y, patient_ids, icu_hours,
                     seq_length=12, stride=6):

    X_seq, y_seq = [], []

    data = pd.DataFrame({
        "patient_id": patient_ids.to_numpy(),
        "icu_hours": icu_hours.to_numpy(),
        "label": y.to_numpy(),
        "position": np.arange(len(X))
    })

    for _, group in data.groupby("patient_id", sort=False):
        group = group.sort_values("icu_hours")
        positions = group["position"].to_numpy()
        labels = group["label"].to_numpy()

        for end in range(seq_length - 1, len(group), stride):
            start = end - seq_length + 1
            window_positions = positions[start:end + 1]

            hours = group["icu_hours"].to_numpy()[start:end + 1]
            if len(hours) != seq_length or np.any(np.diff(hours) != 1):
                continue

            X_seq.append(X[window_positions])
            y_seq.append(labels[end])

    return np.asarray(X_seq, dtype=np.float32), np.asarray(y_seq, dtype=np.int32)


train_hours = df_clean.loc[X_train.index, "ICULOS"]
val_hours = df_clean.loc[X_val.index, "ICULOS"]
test_hours = df_clean.loc[X_test.index, "ICULOS"]

X_train_seq, y_train_seq = create_sequences(
    X_train_scaled, y_train, train_ids, train_hours
)

X_val_seq, y_val_seq = create_sequences(
    X_val_scaled, y_val, val_ids, val_hours
)

X_test_seq, y_test_seq = create_sequences(
    X_test_scaled, y_test, test_ids, test_hours
)

print("Training sequences:", X_train_seq.shape, y_train_seq.shape)
print("Validation sequences:", X_val_seq.shape, y_val_seq.shape)
print("Test sequences:", X_test_seq.shape, y_test_seq.shape)

print("Training positive rate:", y_train_seq.mean())
print("Validation positive rate:", y_val_seq.mean())
print("Test positive rate:", y_test_seq.mean())
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.utils.class_weight import compute_class_weight

tf.keras.utils.set_random_seed(42)

classes = np.unique(y_train_seq)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train_seq
)

class_weights = dict(zip(classes, weights))

print("Class weights:", class_weights)

model = Sequential([
    Input(shape=(X_train_seq.shape[1], X_train_seq.shape[2])),
    LSTM(64, return_sequences=True),
    Dropout(0.3),
    LSTM(32),
    Dropout(0.3),
    Dense(16, activation="relu"),
    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=[
        tf.keras.metrics.AUC(curve="PR", name="auprc"),
        tf.keras.metrics.AUC(curve="ROC", name="roc_auc"),
        tf.keras.metrics.Recall(name="recall")
    ]
)

callbacks = [
    EarlyStopping(
        monitor="val_auprc",
        mode="max",
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor="val_auprc",
        mode="max",
        factor=0.5,
        patience=2,
        min_lr=1e-6,
        verbose=1
    )
]

history = model.fit(
    X_train_seq,
    y_train_seq,
    validation_data=(X_val_seq, y_val_seq),
    epochs=40,
    batch_size=256,
    class_weight=class_weights,
    callbacks=callbacks,
    verbose=1
)
```

```python
import joblib
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)




y_test_prob = model.predict(X_test_seq, verbose=1).ravel()
y_test_pred = (y_test_prob >= 0.5).astype(int)

print("\n========== LSTM TEST RESULTS ==========")
print(f"Accuracy:  {accuracy_score(y_test_seq, y_test_pred):.4f}")
print(f"Precision: {precision_score(y_test_seq, y_test_pred, zero_division=0):.4f}")
print(f"Recall:    {recall_score(y_test_seq, y_test_pred, zero_division=0):.4f}")
print(f"F1-score:  {f1_score(y_test_seq, y_test_pred, zero_division=0):.4f}")
print(f"AUPRC:     {average_precision_score(y_test_seq, y_test_prob):.4f}")
print(f"ROC-AUC:   {roc_auc_score(y_test_seq, y_test_prob):.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test_seq,
    y_test_pred,
    target_names=["Non-Sepsis", "Sepsis"],
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(y_test_seq, y_test_pred))

```
